```python
# ÁRVORE BINÁRIA DE BUSCA
# Representação usando dicionário

arvore = {}

valores = [53, 30, 72, 14, 39]

for valor in valores:

    print("\n========================")
    print(f"Inserindo valor: {valor}")

    # Se a árvore estiver vazia
    if len(arvore) == 0:

        arvore[valor] = {"esquerda": None, "direita": None}
        raiz = valor

        print(f"{valor} virou a raiz")

    else:

        atual = raiz

        print(f"Começando pela raiz: {atual}")

        while True:

            print(f"\nEstou no nó: {atual}")

            # VALOR MENOR -> ESQUERDA
            if valor < atual:

                print(f"{valor} < {atual} → ir para ESQUERDA")

                # Se esquerda estiver vazia
                if arvore[atual]["esquerda"] is None:

                    print(f"Esquerda de {atual} está vazia")

                    arvore[atual]["esquerda"] = valor

                    arvore[valor] = {
                        "esquerda": None,
                        "direita": None
                    }

                    print(f"{valor} inserido à esquerda de {atual}")

                    break

                else:

                    print(f"Esquerda NÃO está vazia")
                    print(f"Descendo para {arvore[atual]['esquerda']}")

                    atual = arvore[atual]["esquerda"]

            # VALOR MAIOR -> DIREITA
            else:

                print(f"{valor} >= {atual} → ir para DIREITA")

                # Se direita estiver vazia
                if arvore[atual]["direita"] is None:

                    print(f"Direita de {atual} está vazia")

                    arvore[atual]["direita"] = valor

                    arvore[valor] = {
                        "esquerda": None,
                        "direita": None
                    }

                    print(f"{valor} inserido à direita de {atual}")

                    break

                else:

                    print(f"Direita NÃO está vazia")
                    print(f"Descendo para {arvore[atual]['direita']}")

                    atual = arvore[atual]["direita"]

print("\n========================")
print("Árvore final:")
print(arvore)
```
