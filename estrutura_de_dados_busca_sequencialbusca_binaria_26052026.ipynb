```python
def busca_sequencial(lista, valor):
    for i in range(len(lista)):
        if lista[i] == valor:
            return i
    return -1


# Exemplo de uso
dados = [8, 3, 15, 10, 21]
resultado = busca_sequencial(dados, 10)

if resultado != -1:
    print(f"Valor encontrado na posição {resultado}")
else:
    print("Valor não encontrado")
```

```python
def busca_binaria(lista, valor):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2

        if lista[meio] == valor:
            return meio
        elif valor < lista[meio]:
            fim = meio - 1
        else:
            inicio = meio + 1

    return -1


# Exemplo de uso
dados = [2, 5, 8, 12, 16, 20, 25]
resultado = busca_binaria(dados, 16)

if resultado != -1:
    print(f"Valor encontrado na posição {resultado}")
else:
    print("Valor não encontrado")
```
