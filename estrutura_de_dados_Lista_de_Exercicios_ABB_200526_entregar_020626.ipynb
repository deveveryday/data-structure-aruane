```python
#Exercício 1 — Construção da Árvore
#Considere a lista abaixo:

#valores = [53, 30, 72, 14, 39, 61, 84, 9, 23, 34, 49, 79]
#Implemente o algoritmo de inserção em uma árvore binária de busca utilizando dicionários.
#Cada nó deve possuir:

#{"esquerda": None, "direita": None}
#Ao final:
#Mostre o dicionário completo da árvore.
#Informe qual é a raiz.
#Explique por que cada valor foi para esquerda ou direita.

# Resposta: No código e: raiz 53. 30 foi para esquerda por ser menor que a raiz, 72 foi para direita por ser maior.
# Em seguida o 14 seguiu para esquerda por ser menor que a nova raiz 30, indo assim até o 79
```

```python
def insert_in_tree(tree, root, value):
  if not root:
      tree[value] = {"left" : None, "right" : None}
      return value
  current = root

  while True:
    if value < current:
        if tree[current]["left"] is None:
            tree[current]["left"] = value
            tree[value] = {"left" : None, "right" : None}
            break
        else:
            current = tree[current]["left"]
    else:
        if tree[current]["right"] is None:
            tree[current]["right"] = value
            tree[value] = {"left" : None, "right" : None}
            break
        else:
            current = tree[current]["right"]

  return root


tree = {}
values = [53, 30, 72, 14, 39, 61, 84, 9, 23, 34, 49, 79]
source = None

for value in values:
    source = insert_in_tree(tree, source, value)


#Mostre o dicionário completo da árvore.
print(tree)

#Informe qual é a raiz.
print(list(tree)[0])

# Resposta: é o 53

#Explique por que cada valor foi para esquerda ou direita.

# Resposta: No código e inicialmente o 30 foi para esquerda por ser menor que a raiz 53, o 72 para direita por ser maior.
# Posteriormente o 14 segue para esquerda por ser menor que a nova raiz 30, assim segue até o 79 que seguir para direita
```

```python
def search_value(tree, root, value_to_find):
    path = []
    current = root
    founded = False
    count_visited = 0

    while current is not None:
        path.append(current)
        count_visited += 1

        if value_to_find == current:
            founded = True
            break
        elif value_to_find < current:
            if current in tree:
                current = tree[current]["left"]
            else:
                current = None
        else:
            if current in tree:
                current = tree[current]["right"]
            else:
                current = None

    return founded, path, count_visited
```

```python
def preorder_traversal(tree, root, result = None):
    if result is None:
        result = []
    if root is not None:
        result.append(root)
        if root in tree:
            preorder_traversal(tree, tree[root]["left"], result)
            preorder_traversal(tree, tree[root]["right"], result)
    return result


print(preorder_traversal(tree, source))
```

```python
#Exercício 7 — Altura da Árvore
#Implemente um algoritmo que calcule a altura da árvore.
#Definição:
#A altura é o maior número de níveis da árvore.
#Seu programa deve:
#Informar a altura da árvore.
#Mostrar qual caminho gera essa altura.

# Resposta: No código e: altura 4
```

```python
from collections import deque

def get_tree_levels(tree, root):
  if root is None or root not in tree and not tree:
    return []

  levels = []
  queue = deque([(root, 0)])

  while queue:
    current_node, level = queue.popleft()

    if len(levels) == level:
      levels.append([])
    levels[level].append(current_node)

    if current_node in tree:
      left_child = tree[current_node]["left"]
      right_child = tree[current_node]["right"]

      if left_child is not None and left_child in tree:
        queue.append((left_child, level + 1))
      if right_child is not None and right_child in tree:
        queue.append((right_child, level + 1))

  return levels
```

```python
#Exercício 13 — Busca de Profundidade
#Implemente uma busca em profundidade (DFS) utilizando pilha.
#Seu algoritmo deve:
#Mostrar a ordem dos nós visitados.
#Procurar um valor específico.
#Informar em qual nível o valor foi encontrado

# Resposta: No código e:
# 49: encontrado no nível 4 (3), descendo os níveis. Ordem: [53, 30, 14, 9, 23, 39, 34, ...]

#               53
#            /      \
#          30        72
#        /   \      /   \
#      14    39   61    84
#     / \    / \         /
#    9  23 34 49       79
```

```python
def dfs_tree_search(tree, root, target_value):
  if not tree or root is None:
    return False, [], -1

  stack = [(root, [root], 0)]
  visited_nodes_order = []

  while stack:
    current_node, current_path, current_level = stack.pop()
    visited_nodes_order.append(current_node)

    if current_node == target_value:
      print(f"Value {target_value} found")
      print(f"Order of nodes visited (DFS): {visited_nodes_order}")
      print(f"Path to target: {current_path}")
      print(f"Found at level: {current_level}")
      return True, visited_nodes_order, current_level

    if current_node in tree:
      right_child = tree[current_node]["right"]
      left_child = tree[current_node]["left"]

      if right_child is not None and right_child in tree:
        stack.append((right_child, current_path + [right_child], current_level + 1))

      if left_child is not None and left_child in tree:
        stack.append((left_child, current_path + [left_child], current_level + 1))

  print(f"Value {target_value} not found.")
  print(f"Order DFS: {visited_nodes_order}")

  return False, visited_nodes_order, -1


dfs_tree_search(tree, source, 49)
```

```python
#Exercício 14 — Busca em Largura
#Implemente uma busca em largura (BFS) utilizando fila.
#Seu programa deve:
#Mostrar os nós nível por nível.
#Informar quantos níveis existem.
#Comparar BFS com DFS.

# Resposta: No código e: BFS literalmente faz fila, DFS faz pilha
# Level 0: [53]
# Level 1: [30, 72]
# Level 2: [14, 39, 61, 84]
# Level 3: [9, 23, 34, 49, 79]
```

```python
from collections import deque

def bfs_traverse_and_search(tree, root, target_value=None):
  if not tree or root is None:
    return [], [], -1, False

  queue = deque([(root, 0, [root])])
  visited_nodes_order = []
  levels = []
  found = False
  target_level = -1
  target_path = []

  while queue:
    current_node, current_level, current_path = queue.popleft()
    visited_nodes_order.append(current_node)

    if target_value is not None and current_node == target_value:
      found = True
      target_level = current_level
      target_path = current_path

    while len(levels) <= current_level:
      levels.append([])

    levels[current_level].append(current_node)

    if current_node in tree:
      left_child = tree[current_node]["left"]
      right_child = tree[current_node]["right"]

      if left_child is not None and left_child in tree:
        queue.append((left_child, current_level + 1, current_path + [left_child]))

      if right_child is not None and right_child in tree:
        queue.append((right_child, current_level + 1, current_path + [right_child]))

  num_levels = len(levels)

  return visited_nodes_order, levels, num_levels, found, target_level, target_path



print(tree)

visited_bfs, tree_levels_bfs, num_levels_bfs, found_bfs, level_bfs, path_bfs = bfs_traverse_and_search(tree, source, 61)

print("\nNodes visited in BFS order:", visited_bfs)
print("Tree Levels (BFS):")
for i, level_nodes in enumerate(tree_levels_bfs):
    print(f"Level {i}: {level_nodes}")
print(f"Total number of levels: {num_levels_bfs}")

if found_bfs:
    print(f"{level_bfs}, Path: {path_bfs}")
else:
    print("Value not found.")
```

```python
#Exercício 15 — Verificação de Existência
#Crie uma função:
#existe(arvore, valor)
#A função deve retornar:
#True → se o valor existir
#False → caso contrário
#Teste para:
#23
#100
#61

# Resposta: No código
```

```python
def exists(tree, root, value_to_check):
  if not tree or root is None:
    return False

  current = root

  while current is not None:
    if value_to_check == current:
      return True
    elif value_to_check < current:
      if current in tree:
        current = tree[current]["left"]
      else:
        return False
    else:
      if current in tree:
        current = tree[current]["right"]
      else:
        return False

  return False

print(tree)

print(f"{exists(tree, source, 23)}")

print(f"{exists(tree, source, 100)}")

print(f"{exists(tree, source, 61)}")
```

```python
def calculate_tree_sum_and_average(tree):
  if not tree:
    return 0, 0

  total_sum = 0
  node_count = 0
  node_values = []

  for node_val in tree.keys():
    node_values.append(node_val)
    total_sum += node_val
    node_count += 1

  average = total_sum / node_count if node_count > 0 else 0

  return total_sum, average

total, the_avg = calculate_tree_sum_and_average(tree)

print(f"Soma: {total}. Média: {the_avg}")
```

```python
def check_tree_balanced(tree, node):
  if node is None or node not in tree:
    return True, -1

  left_balanced, left_height = check_tree_balanced(tree, tree[node]["left"])

  if not left_balanced:
    return False, 0

  right_balanced, right_height = check_tree_balanced(tree, tree[node]["right"])

  if not right_balanced:
    return False, 0

  if abs(left_height - right_height) > 1:
    return False, 0

  return True, 1 + max(left_height, right_height)


print("Balanceada:", check_tree_balanced(tree, source)[0])
```

```python
#DESAFIO
#Árvores AVL

# Resposta: São árvores que mantém balanceamento visando performance, de forma que lados esquerdo e direito fiquem com quantidade semelhante de registros
```

```python
def create_node(data):
  return {"data" : data, "left" : None, "right" : None, "height" : 1}


def get_height(node):
  if node is None:
    return 0
  return node["height"]


def get_balance(node):
  if node is None:
    return 0
  return get_height(node["left"]) - get_height(node["right"])


def right_rotate(y):
  print("right ", y["data"])

  x = y["left"]
  T2 = x["right"]

  x["right"] = y
  y["left"] = T2

  y["height"] = 1 + max(get_height(y["left"]), get_height(y["right"]))
  x["height"] = 1 + max(get_height(x["left"]), get_height(x["right"]))

  return x


def left_rotate(x):
  print("left ", x["data"])

  y = x["right"]
  T2 = y["left"]

  y["left"] = x
  x["right"] = T2

  x["height"] = 1 + max(get_height(x["left"]), get_height(x["right"]))
  y["height"] = 1 + max(get_height(y["left"]), get_height(y["right"]))
  return y

def insert(node, data):
  if node is None:
    return create_node(data)

  if data < node["data"]:
    node["left"] = insert(node["left"], data)
  elif data > node["data"]:
    node["right"] = insert(node["right"], data)
    return node

  node["height"] = 1 + max(get_height(node["left"]), get_height(node["right"]))

  balance = get_balance(node)

  if balance > 1 and get_balance(node["left"]) >= 0:
    return right_rotate(node)

  if balance > 1 and get_balance(node["left"]) < 0:
    node["left"] = left_rotate(node["left"])
    return right_rotate(node)

  if balance < -1 and get_balance(node["right"]) <= 0:
    return left_rotate(node)

  if balance < -1 and get_balance(node["right"]) > 0:
    node["right"] = right_rotate(node["right"])
    return left_rotate(node)

  return node


def inorder_traversal_avl(node, result=None):
  if result is None:
    result = []

  if node is not None:
    inorder_traversal_avl(node["left"], result)
    result.append(node["data"])
    inorder_traversal_avl(node["right"], result)

  return result


root = None
letters = ["C", "B", "E", "A", "D", "H", "G", "F"]

for letter in letters:
  root = insert(root, letter)

print(inorder_traversal_avl(root))
```

```python
#Árvores Rubro-Negras

# Resposta: São árvores que utilizam representação adicional vermelha ou preta, para manter balanceamento em todas operações, para deletar, inserir...
```

```python
RED = "RED"
BLACK = "BLACK"

NIL = {"data" : None, "left" : None, "right" : None, "color" : BLACK, "parent" : None}
NIL["left"] = NIL
NIL["right"] = NIL

def create_rb_tree_node(data):
  return {"data" : data, "left" : NIL, "right" : NIL, "color" : RED, "parent" : NIL}


def get_rb_tree_color(node):
  if node is NIL:
    return BLACK

  return node["color"]


def set_rb_tree_color(node, color):
  if node is not NIL:
    node["color"] = color


def rb_tree_left_rotate(tree_root, x):
  y = x["right"]
  x["right"] = y["left"]

  if y["left"] is not NIL:
    y["left"]["parent"] = x

  y["parent"] = x["parent"]
  if x["parent"] is NIL:
    tree_root = y
  elif x == x["parent"]["left"]:
    x["parent"]["left"] = y
  else:
    x["parent"]["right"] = y
  y["left"] = x
  x["parent"] = y

  return tree_root


def rb_tree_right_rotate(tree_root, y):
  x = y["left"]
  y["left"] = x["right"]
  if x["right"] is not NIL:
    x["right"]["parent"] = y

  x["parent"] = y["parent"]
  if y["parent"] is NIL:
    tree_root = x
  elif y == y["parent"]["right"]:
    y["parent"]["right"] = x
  else:
    y["parent"]["left"] = x
  x["right"] = y
  y["parent"] = x
  return tree_root


def rb_tree_insert_fixup(tree_root, z):
  while z["parent"]["color"] == RED:
    if z["parent"] == z["parent"]["parent"]["left"]:
      y = z["parent"]["parent"]["right"]

      if y["color"] == RED:
        set_rb_tree_color(z["parent"], BLACK)
        set_rb_tree_color(y, BLACK)
        set_rb_tree_color(z["parent"]["parent"], RED)
        z = z["parent"]["parent"]
      else:
        if z == z["parent"]["right"]:
          z = z["parent"]
          tree_root = rb_tree_left_rotate(tree_root, z)

        set_rb_tree_color(z["parent"], BLACK)
        set_rb_tree_color(z["parent"]["parent"], RED)
        tree_root = rb_tree_right_rotate(tree_root, z["parent"]["parent"])
    else:
      y = z["parent"]["parent"]["left"]
      if y["color"] == RED:
        set_rb_tree_color(z["parent"], BLACK)
        set_rb_tree_color(y, BLACK)
        set_rb_tree_color(z["parent"]["parent"], RED)
        z = z["parent"]["parent"]
      else:
        if z == z["parent"]["left"]:
          z = z["parent"]
          tree_root = rb_tree_right_rotate(tree_root, z)

        set_rb_tree_color(z["parent"], BLACK)
        set_rb_tree_color(z["parent"]["parent"], RED)
        tree_root = rb_tree_left_rotate(tree_root, z["parent"]["parent"])

  set_rb_tree_color(tree_root, BLACK)

  return tree_root


def rb_tree_insert(tree_root, data):
  z = create_rb_tree_node(data)
  y = NIL
  x = tree_root

  while x is not NIL:
    y = x
    if z["data"] < x["data"]:
      x = x["left"]
    elif z["data"] > x["data"]:
      x = x["right"]
    else:
      return tree_root

  z["parent"] = y
  if y is NIL:
    tree_root = z
  elif z["data"] < y["data"]:
    y["left"] = z
  else:
    y["right"] = z

  tree_root = rb_tree_insert_fixup(tree_root, z)

  return tree_root


def rb_tree_inorder_traversal(node, result = None):
  if result is None:
    result = []
  if node is not NIL:
    rb_tree_inorder_traversal(node["left"], result)
    result.append(f"{node['data']}({node['color']})")
    rb_tree_inorder_traversal(node["right"], result)

  return result


rb_root = NIL
letters = ["C", "B", "E", "A", "D", "H", "G", "F", "I", "J"]

for letter in letters:
  rb_root = rb_tree_insert(rb_root, letter)
  current_rb_tree_state = rb_tree_inorder_traversal(rb_root)

print(' '.join(rb_tree_inorder_traversal(rb_root)))
```
